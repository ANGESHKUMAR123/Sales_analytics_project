```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
print("Libraries loaded successfully.")
```

```python
file_path="data/raw_sales_data_2014_2025_practice.csv"
df=pd.read_csv(file_path)
print("rows:",len(df))
print("column:",len(df.columns))
df
```

```python
print("shape")
print(df.shape)
print("\n column names")
print(df.columns.tolist())
print("\ndata types")
print(df.dtypes)
print("\ninfo")
print(df.info())
print("\nfirst_10 value")
display(df.head(10))
print("\nlast_10 value")
display(df.tail(10))
print("\n random sample")
display(df.sample(10,random_state=42))
```

```python
print("numeric summary")
display(df.describe().T)
print("categorical  summary")
display(df.describe(include="object").T)
```

```python
quality=pd.DataFrame({
    "missing_count":df.isna().sum(),
    "missing_percent":df.isna().mean()*100,
    "unique_value":df.nunique(dropna=True)
}).sort_values("missing_count",ascending=False)
display(quality)
print("duplicate_rows:",df.duplicated().sum())
print("quantity<=0:",(df["quantity"]<=0) .sum())
print("unit_price<=0:",(df["unit_price"]<=0).sum())
print("discount above 0-1:",((df["discount"]<0) |(df["discount"]>1)).sum())
print("data parse failure:",pd.to_datetime(df["order_date"],errors="coerce").isna().sum())

```

```python
raw_df = df.copy()
clean = df.copy()
print("Raw rows:", len(raw_df))
print("Working rows:", len(clean))
```

```python
clean.columns = (
 clean.columns
 .str.strip()
 .str.lower()
 .str.replace(" ", "_", regex=False)
 .str.replace(r"[^a-z0-9_]", "", regex=True)
)
print(clean.columns.tolist())
```

```python
clean["order_date"]=pd.to_datetime(df["order_date"],errors="coerce")
numeric_columns=["order_id","customer_id","quantity","unit_price","discount","sales","cost","profit"]
for col in numeric_columns:
    clean[col]=pd.to_numeric(df[col],errors="coerce")
print(clean.dtypes)
```

```python
text_col=["region","category","channel"]
for text in text_col:
    clean[text]=clean[text].astype("string").str.strip()
clean["region"]=clean["region"].str.title()
clean["category"]=clean["category"].str.title()
clean["channel"]=clean["channel"].str.title()
display(clean[text_col].drop_duplicates().sort_values(text_col))
```

```python
display(clean.head(10))
```

```python
duplicates_before=clean.duplicated().sum()
clean=clean.drop_duplicates().copy()
duplicates_after=clean.duplicated().sum()
print("dulpicate_before:",duplicates_before)
print("duplicate_after:",duplicates_after)
print("numder rows after remove dupicate:",len(clean))
```

```python
print("missing count order_id:",clean["order_id"].isna().sum())
print("missing duplicates order_id:",clean["order_id"].duplicated().sum())
duplicate_order_id=clean[clean["order_id"].duplicated(keep=False)]
display(duplicate_order_id.head())
```

```python
invalid_quantity=clean["quantity"]<=0
invalid_unit_price=clean["unit_price"]<=0
print("invalid_quantity:",invalid_quantity.sum())
print("invalid_unit_price:",invalid_unit_price.sum())
clean.loc[invalid_quantity,"quantity"]=np.nan
clean.loc[invalid_unit_price,"unit_price"]=np.nan
clean["quantity"]=clean["quantity"].fillna(clean["quantity"].median())
clean["unit_price"]=clean["unit_price"].fillna(clean["unit_price"].median())
```

```python
invalid_discount=((clean["discount"]<0)|(clean["discount"]>1))
print("invalid_discount:",invalid_discount.sum())
```

```python
print("missing_discount:",clean["discount"].isna().sum())
```

```python
clean.loc[invalid_discount,"discount"]=np.nan
clean["discount"]=clean["discount"].fillna(clean["discount"].median())
print("after clean discount:",clean["discount"].isna().sum())
```

```python
for col in ["region", "category", "channel"]:
 clean[col] = clean[col].fillna("Unknown")
print(clean[["region", "category", "channel"]].isna().sum())
```

```python
clean["sales"]=(clean["quantity"]*clean["unit_price"]*(1-clean["discount"]))
clean["cost"]=clean["sales"]*0.70
clean["profit"] = clean["sales"] - clean["cost"]
clean["profit_megin"]=np.where(clean["sales"]!=0,clean["profit"]/clean["sales"]*100,0)
display(clean.head(10))
```

```python
clean["year"]=clean["order_date"].dt.year
clean["month"]=clean["order_date"].dt.month
clean["quarter"]="Q"+ clean["order_date"].dt.quarter.astype(str)
clean["month_name"]=clean["order_date"].dt.strftime("%b")
clean["year_month"]=(clean["order_date"].dt.to_period("M").astype(str))
clean["days"]=clean["order_date"].dt.strftime("%A")
display(clean[
    ["order_date",
     "quarter",
     "year",
     "month",
     "month_name",
     "year_month",
     "days"
    ]].head())
```

```python
display(clean)
```

```python
print("missing_value")
display(clean.isna().sum().sort_values(ascending=False))
print("invlide_quantity:",(clean["quantity"]<=0).sum())
print("invlide_unit_price:",(clean["unit_price"]<=0).sum())
print("invlide_discount:",((clean["discount"]<0)|(clean["discount"]>1)).sum())
print("nagivate_sales:",(clean["sales"]<0).sum())
print("infinite_value:",np.isinf(clean.select_dtypes(include=np.number)).sum().sum())
assert clean["order_date"].notna().all()
assert (clean["quantity"]>0).all()
assert (clean["unit_price"]>0).all()
assert (clean["sales"]>0).all()
assert (clean["discount"].between(0,1)).all()
print("all validation test is pass✅")
```

```python
comparison = pd.DataFrame({
 "metric": [
 "Rows",
 "Columns",
 "Duplicate rows",
 "Missing cells"
 ],
 "raw": [
 len(raw_df),
 len(raw_df.columns),
 raw_df.duplicated().sum(),
 raw_df.isna().sum().sum()
 ],
 "clean": [
 len(clean),
 len(clean.columns),
 clean.duplicated().sum(),
 clean.isna().sum().sum()
 ]
})
display(comparison)
```

```python
import os
os.makedirs("output", exist_ok=True)
clean_path = "output/clean_sales_data.csv"
clean.to_csv(clean_path, index=False)
print("Saved:", clean_path)
print("Rows:", len(clean))
```

```python
total_sales=clean["sales"].sum()
total_profit=clean["profit"].sum()
total_quantity=clean["quantity"].sum()
total_orderes=clean["order_id"].nunique() 
profit_mergin=total_profit/total_sales*100
average_order_value=total_sales/total_orderes
kpis=pd.Series({
    "Total Sales:":total_sales,
    "Total Profit:":total_profit,
    "Total Quantity:":total_quantity,
    "Total orders:":total_orderes,
    "Profit Mergin:":profit_mergin,
    "Average Order value:":average_order_value
})
display(kpis)
```

```python
region_summary=(
    clean.groupby("region",as_index=False)
        .agg(
            sales=("sales","sum"),
            orders=("order_id","nunique"),
            profit=("profit","sum"),
            quantity=("quantity","sum")
))
region_summary["profit_mergin"]=(region_summary["profit"]/region_summary["sales"]*100)
region_summary=region_summary.sort_values("sales",ascending=False)
display(region_summary)
```

```python
category_summary=(
    clean.groupby("category",as_index=False)
    .agg(
        sales=("sales","sum"),
        profit=("profit","sum"),
        quantity=("quantity","sum"),
        orders=("order_id","sum")
    )
)
category_summary["profit_mergin"]=(category_summary["profit"]/category_summary["sales"]*100)
display(category_summary.sort_values("sales",ascending=False))
```

```python
channel_summary=(
    clean.groupby("channel",as_index=False)
    .agg(
        sales=("sales","sum"),
        profit=("profit","sum"),
        quantity=("quantity","sum"),
        orders=("order_id","sum")
    )
)
channel_summary["profit_mergin"]=(channel_summary["profit"]/channel_summary["sales"])
display(channel_summary.sort_values("sales",ascending=False))
```

```python
monthly_summary = (
 clean.groupby("year_month", as_index=False)
 .agg(
 sales=("sales", "sum"),
 profit=("profit", "sum"),
 orders=("order_id", "nunique")
 )
 .sort_values("year_month")
)
monthly_summary["profit_mergin"] = (
 monthly_summary["profit"] /
 monthly_summary["sales"]*100
)
display(monthly_summary.head(12))
```

```python
plt.figure(figsize=(14, 5))
plt.plot(monthly_summary["year_month"],
 monthly_summary["sales"])
plt.xticks(rotation=60)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.bar(category_summary["category"],
 category_summary["profit_mergin"])
plt.xticks(rotation=20)
plt.title("Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin")
plt.tight_layout()
plt.show()
```

```python
customer_summary=(clean.groupby("customer_id",as_index=False).agg(
    sales=("sales","sum"),
    profit=("profit","sum"),
    orders=("order_id","nunique")
))
top_customer=customer_summary.sort_values("sales",ascending=False).head(10)
display(top_customer)
```

```python
region_summary.to_csv("output/region_summary.csv",index=False)
category_summary.to_csv("output/category_summary.csv",index=False)
channel_summary.to_csv("output/channel_summary.csv",index=False)
monthly_summary.to_csv("output/monthly_summary.csv",index=False)
top_customer.to_csv("output/top_customer.csv",index=False)
print("All file are created successfully")
```
